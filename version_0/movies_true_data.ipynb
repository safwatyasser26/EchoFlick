```python
movies_data = {}
movie_schema = {}
```

```python
def store(id,results):
    movie_schema['imdb_id'] = id
    movie_schema['tmdb_id'] = results['id']
    keys = ['overview','poster_path','popularity','release_date']
    for key in keys:
        movie_schema[key] = results[key] 
    movies_data[id] = dict(movie_schema.items())
    print(round((len(movies_data[id]) / len(movie_ids)) * 100,2))
    (
     pd.DataFrame()
     .from_dict(movies_data)
     .to_json('movies_tmdb_data.json')
    )
```

```python
def get_movies_data(ids):
    for movie_id in ids:
        data = req.get(find_by_id_url.replace('external_id',movie_id), headers=headers).json()
        movie_results = data['movie_results']
        if len(movie_results) > 0:
            store(movie_id,dict(movie_results[0].items()))
```
